```python
import json
from pathlib import Path

SWEEP   = Path("/workspace/food101_lt/sweep")
POOL_F  = Path("/workspace/food101_lt_synth")
SWEEP_M = json.loads((SWEEP/"manifest.json").read_text())

pool_total = sum(len(list(d.glob("*.jpg"))) for d in POOL_F.iterdir() if d.is_dir())
print(f"pool holds {pool_total} images\n")

for m in sorted(SWEEP_M["ratios"], key=lambda x: -x["imbalance_ratio"]):
    ir, real = m["imbalance_ratio"], m["total_train_images"]
    line = f"IR {ir:>3}  real {real:>6}  "
    for frac in (0.25, 0.15, 0.10, 0.05):
        need = int(real * frac)
        line += f"| {frac:.0%}:{need:>5}{'ok' if need <= pool_total else 'SHORT'} "
    print(line)
```

```python
FIXED_BUDGET = 3800

print(f"{'IR':>4} {'real':>7} {'synth':>6} {'frac':>6}")
for m in sorted(SWEEP_M["ratios"], key=lambda x: -x["imbalance_ratio"]):
    ir, real = m["imbalance_ratio"], m["total_train_images"]
    print(f"{ir:>4} {real:>7} {FIXED_BUDGET:>6} {FIXED_BUDGET/real:>5.1%}")
```
